# Read ifdd-be with your own rights

`ifdd_lab.session()` sends the access token of the person logged in to the hub with every request. ifdd-be answers with exactly what your roles and context patterns allow: a domain context without a pattern gives 403.

Never print the token: outputs are saved with the notebook.

```python
import ifdd_lab

be = ifdd_lab.session()   # base URL: IFDD_BE_URL
who = be.get("/whoami")
who.raise_for_status()
me = who.json()
print("logged in as:", me["preferred_username"])
print("roles:", me["client_roles"])
print("context patterns:", me["context_patterns"])
```

```python
r = be.get("/nodes/", params={"start": 0, "end": 5})
print(r.status_code, "- objects visible to you:", r.headers.get("X-Total-Count"))
for item in (r.json() if r.ok else []):
    print(item.get("@id"))
```
